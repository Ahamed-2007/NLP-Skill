```python
# ============================================================
# CUSTOMER REVIEW CLEANING SYSTEM - IMDb 50K DATASET
# ============================================================


# ------------------------------------------------------------
# STEP 1: IMPORT LIBRARIES
# ------------------------------------------------------------

import pandas as pd
import re
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer


# ------------------------------------------------------------
# STEP 2: DOWNLOAD NLTK DATA
# ------------------------------------------------------------

nltk.download("stopwords")
nltk.download("punkt")
nltk.download("punkt_tab")


# ------------------------------------------------------------
# STEP 3: LOAD IMDb DATASET
# ------------------------------------------------------------

df = pd.read_csv("IMDB Dataset.csv")

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns)

print("\nFirst 5 rows:")
print(df.head())


# ------------------------------------------------------------
# STEP 4: CHECK MISSING VALUES
# ------------------------------------------------------------

print("\nMissing values:")
print(df.isnull().sum())


# ------------------------------------------------------------
# STEP 5: CHECK AND REMOVE DUPLICATES
# ------------------------------------------------------------

print("\nDuplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)


# ------------------------------------------------------------
# STEP 6: CONVERT TEXT TO LOWERCASE
# ------------------------------------------------------------

df["cleaned_review"] = df["review"].str.lower()


# ------------------------------------------------------------
# STEP 7: REMOVE HTML TAGS
# ------------------------------------------------------------

df["cleaned_review"] = df["cleaned_review"].apply(
    lambda x: re.sub(r"<.*?>", "", x)
)


# ------------------------------------------------------------
# STEP 8: REMOVE URLs
# ------------------------------------------------------------

df["cleaned_review"] = df["cleaned_review"].apply(
    lambda x: re.sub(r"http\S+|www\S+", "", x)
)


# ------------------------------------------------------------
# STEP 9: REMOVE NUMBERS
# ------------------------------------------------------------

df["cleaned_review"] = df["cleaned_review"].apply(
    lambda x: re.sub(r"\d+", "", x)
)


# ------------------------------------------------------------
# STEP 10: REMOVE PUNCTUATION AND SPECIAL CHARACTERS
# ------------------------------------------------------------

df["cleaned_review"] = df["cleaned_review"].apply(
    lambda x: re.sub(r"[^a-zA-Z\s]", "", x)
)


# ------------------------------------------------------------
# STEP 11: REMOVE EXTRA SPACES
# ------------------------------------------------------------

df["cleaned_review"] = df["cleaned_review"].apply(
    lambda x: re.sub(r"\s+", " ", x).strip()
)


# ------------------------------------------------------------
# STEP 12: LOAD STOPWORDS
# ------------------------------------------------------------

stop_words = set(stopwords.words("english"))

print("\nNumber of English stopwords:", len(stop_words))


# ------------------------------------------------------------
# STEP 13: TOKENIZATION + STOPWORD REMOVAL
# ------------------------------------------------------------

def tokenize_and_remove_stopwords(text):

    # Convert sentence into words
    tokens = word_tokenize(text)

    # Remove stopwords
    tokens = [
        word for word in tokens
        if word not in stop_words
    ]

    return tokens


df["tokens"] = df["cleaned_review"].apply(
    tokenize_and_remove_stopwords
)


# ------------------------------------------------------------
# STEP 14: STEMMING
# ------------------------------------------------------------

stemmer = PorterStemmer()

df["stemmed_tokens"] = df["tokens"].apply(
    lambda tokens: [
        stemmer.stem(word)
        for word in tokens
    ]
)


# ------------------------------------------------------------
# STEP 15: CONVERT STEMMED TOKENS BACK TO TEXT
# ------------------------------------------------------------

df["cleaned_review"] = df["stemmed_tokens"].apply(
    lambda tokens: " ".join(tokens)
)


# ------------------------------------------------------------
# STEP 16: COMPARE ORIGINAL AND CLEANED REVIEWS
# ------------------------------------------------------------

print("\n====================================================")
print("ORIGINAL VS CLEANED REVIEWS")
print("====================================================")

for i in range(5):

    print("\nOriginal Review:")
    print(df["review"].iloc[i])

    print("\nCleaned Review:")
    print(df["cleaned_review"].iloc[i])

    print("\n" + "-" * 80)


# ------------------------------------------------------------
# STEP 17: DISPLAY FINAL DATASET
# ------------------------------------------------------------

print("\nFinal Dataset:")
print(df.head())


# ------------------------------------------------------------
# STEP 18: CHECK FINAL COLUMNS
# ------------------------------------------------------------

print("\nFinal Columns:")
print(df.columns)


# ------------------------------------------------------------
# STEP 19: EXPORT PREPROCESSED DATASET
# ------------------------------------------------------------

df.to_csv(
    "IMDB_Preprocessed.csv",
    index=False
)

print("\n====================================================")
print("SUCCESS!")
print("Preprocessed dataset saved as:")
print("IMDB_Preprocessed.csv")
print("====================================================")
```
