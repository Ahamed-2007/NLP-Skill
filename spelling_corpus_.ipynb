```python
# ============================================================
# TASK 2: SEARCH QUERY SPELLING CORRECTOR
# ============================================================

import re
import nltk

from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")


# ============================================================
# STEP 1: LOAD SPELLING CORPUS
# ============================================================

with open("spelling_corpus.txt", "r", encoding="utf-8") as file:
    corpus_text = file.read()


# ============================================================
# STEP 2: BUILD VOCABULARY
# ============================================================

corpus_text = corpus_text.lower()

vocabulary = set(
    re.findall(r"[a-z]+", corpus_text)
)

print("Vocabulary size:", len(vocabulary))


# ============================================================
# STEP 3: EDIT DISTANCE FUNCTION
# ============================================================

def edit_distance(word1, word2):

    rows = len(word1) + 1
    cols = len(word2) + 1

    matrix = [[0] * cols for _ in range(rows)]

    # First column
    for i in range(rows):
        matrix[i][0] = i

    # First row
    for j in range(cols):
        matrix[0][j] = j

    # Calculate distance
    for i in range(1, rows):

        for j in range(1, cols):

            if word1[i - 1] == word2[j - 1]:
                cost = 0
            else:
                cost = 1

            matrix[i][j] = min(
                matrix[i - 1][j] + 1,
                matrix[i][j - 1] + 1,
                matrix[i - 1][j - 1] + cost
            )

    return matrix[-1][-1]


# ============================================================
# STEP 4: FIND CLOSEST WORD
# ============================================================

def find_correction(word):

    best_word = word
    best_distance = float("inf")

    for candidate in vocabulary:

        distance = edit_distance(word, candidate)

        if distance < best_distance:

            best_distance = distance
            best_word = candidate

    return best_word, best_distance


# ============================================================
# STEP 5: USER INPUT
# ============================================================

query = input("Enter your search query: ")

original_query = query

query = query.lower()

tokens = word_tokenize(query)


# ============================================================
# STEP 6: FIND INCORRECT WORDS
# ============================================================

corrected_tokens = []

suggestions = []


for word in tokens:

    if word.isalpha():

        if word not in vocabulary:

            correction, distance = find_correction(word)

            suggestions.append(
                (word, correction, distance)
            )

            corrected_tokens.append(correction)

        else:

            corrected_tokens.append(word)


# ============================================================
# STEP 7: CREATE CORRECTED QUERY
# ============================================================

corrected_query = " ".join(corrected_tokens)


# ============================================================
# STEP 8: DISPLAY RESULTS
# ============================================================

print("\n==========================================")
print("SPELLING CORRECTION RESULTS")
print("==========================================")

print("\nOriginal Query:")
print(original_query)

print("\nIncorrect Words and Suggested Corrections:")

if len(suggestions) == 0:

    print("No spelling errors found.")

else:

    for wrong, correct, distance in suggestions:

        print(
            wrong,
            "→",
            correct,
            "(Edit Distance:",
            distance,
            ")"
        )

print("\nFinal Corrected Query:")
print(corrected_query)


# ============================================================
# STEP 9: TEST MULTIPLE SPELLING ERRORS
# ============================================================

test_queries = [
    "machne lerning cours",
    "pythn programmng",
    "artifical inteligence",
    "data scince"
]

print("\n==========================================")
print("MULTIPLE QUERY TESTING")
print("==========================================")


for test_query in test_queries:

    tokens = word_tokenize(test_query.lower())

    corrected_tokens = []

    for word in tokens:

        if word.isalpha():

            if word not in vocabulary:

                correction, distance = find_correction(word)

                corrected_tokens.append(correction)

            else:

                corrected_tokens.append(word)

    corrected = " ".join(corrected_tokens)

    print("\nOriginal Query:")
    print(test_query)

    print("Corrected Query:")
    print(corrected)
```
